```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import cross_val_score, GridSearchCV
from sklearn.linear_model import Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

1. Load California Housing dataset

```python
data = fetch_california_housing()
X, y = data.data, data.target
```

2. Create pipelines with standardization

```python
ridge_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('ridge', Ridge())
])

lasso_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('lasso', Lasso(max_iter=2000))
])

# Define alpha values to test
alphas = [0.01, 0.1, 1.0, 10.0, 100.0]
```

3. Comparing L2(Ridge) and L1(Lasso) with GridSearchCV

```python
print("Ridge Regression Results:")
ridge_grid = GridSearchCV(ridge_pipeline, {'ridge__alpha': alphas}, 
                         cv=10, scoring='r2', n_jobs=-1)
ridge_grid.fit(X, y)

print(f"Best alpha: {ridge_grid.best_params_['ridge__alpha']}")
print(f"Best CV score: {ridge_grid.best_score_:.4f}")


print("\nLasso Regression Results:")
lasso_grid = GridSearchCV(lasso_pipeline, {'lasso__alpha': alphas}, 
                         cv=10, scoring='r2', n_jobs=-1)
lasso_grid.fit(X, y)

print(f"Best alpha: {lasso_grid.best_params_['lasso__alpha']}")
print(f"Best CV score: {lasso_grid.best_score_:.4f}")


print(f"\nModel Comparison:")
print(f"Ridge (alpha={ridge_grid.best_params_['ridge__alpha']}): {ridge_grid.best_score_:.4f}")
print(f"Lasso (alpha={lasso_grid.best_params_['lasso__alpha']}): {lasso_grid.best_score_:.4f}")
```

4. Show detailed CV results for best models

```python
print(f"\nDetailed 10-fold CV scores for best Ridge model:")
best_ridge = Pipeline([
    ('scaler', StandardScaler()),
    ('ridge', Ridge(alpha=ridge_grid.best_params_['ridge__alpha']))
])
ridge_scores = cross_val_score(best_ridge, X, y, cv=10, scoring='r2')
print(f"Scores: {ridge_scores}")
print(f"Mean: {ridge_scores.mean():.4f} ± {ridge_scores.std():.4f}")
```
